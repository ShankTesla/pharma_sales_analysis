```python
# import necessary libaries

import pandas as pd
```

```python
# load data into df

df = pd.read_csv("../data/raw/salesdaily.csv", encoding="utf-8", na_values=["", "NA", "-", "?"])
```

```python
# display shape and memory usage

display(df.shape)
display(df.info(memory_usage="deep"))
```

* There is no null values
* There is 2 object data type

```python
# display head and tail

display(df.head())
display(df.tail())
```

* head and tail look normal
* data seems to be of the type MM/DD/YYYY

```python
# display random 10 sample

df.sample(10, random_state=0)
```

```python
# display dtypes and head of dtypes of "object"

print("dtypes")
display(df.dtypes)
print("dtypes head")
display(df.select_dtypes("object").head())
```

* The datum object type seems to be date column MM/DD/YYYY
* The Weekday Name seems to be string representing the day of the week of that date

```python
# display mean of missing values if any

print("Missing value mean if any")
display(df.isna().mean().sort_values(ascending=False))
display(df.isna().any(axis=1).mean())
```

* No missing values found

```python
# Reveal unique IDs, constraints, numerical codes that are categories and high-cardinality categorical data

print("Unique?")
display(df.nunique().sort_values())
```

* Initial pass reveals no duplicate dates

```python
# Summary statistics

print("Summary statistics = any impossible or suspicious values?")
df.describe(percentiles=[.01, .05, .5, .95, .99]).T
```

* Maybe slight right skewness in N02BE, R03
* very slight skewness in rest may or may not be important

```python
# display value counts of categorical columns

print("Categorical columns value counts")

for c in df.select_dtypes("object"):
    print(df[c].value_counts(dropna=False).head())
    print(df[c].value_counts().tail(10))
```

* All seems fair , only wednesday is missing 1 - 300 compared to 301 for every other day

```python
# display duplicate count

print("Duplicate counts")
print(df.duplicated().sum())
print("Duplicate counts for datum column")
df.duplicated(subset=["datum"]).sum()
```

* No duplicates found
