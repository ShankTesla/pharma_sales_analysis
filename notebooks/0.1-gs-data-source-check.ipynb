<h4> Import libraries </h4>

```python
import pandas as pd 
from pathlib import Path 
import hashlib

path = Path("../data/raw/salesdaily.csv")
```

<h4>File Identity</h4>

```python
# Record file identity to detect changes later
print("Size (bytes):", path.stat().st_size)
print("SHA-256:", hashlib.sha256(path.read_bytes()).hexdigest())

df = pd.read_csv(path)
```

<h4> Source claims vs Reality</h4>

```python
# Claims vs Reality: comparing to source
print("Rows, cols:", df.shape)
print("Columns:", list(df.columns))
print("Date range:", df["Year"].min(), " to ", df["Year"].max())
```
