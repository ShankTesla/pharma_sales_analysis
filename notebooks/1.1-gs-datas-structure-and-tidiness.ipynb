```python
# import libraries 

import pandas as pd
import regex
import sys
import re
from pathlib import Path 

PROJECT_ROOT = Path.cwd().parents[0]

if str(PROJECT_ROOT) not in sys.path:
    sys.path.append(str(PROJECT_ROOT))

from pharma_sales_analysis.data.log_utils import log_step
```

```python
# load data into df

data_path = PROJECT_ROOT / "data" / "raw" / "salesdaily.csv"
df = pd.read_csv(data_path, encoding="utf-8", na_values=["", "NA", "-", "?"])
```

```python
# standardise column names

df  = (df.rename(columns=str.strip)
              .pipe(log_step, "01_strip_column_names")
              .rename(columns=str.lower)
              .pipe(log_step, "01_column_names_to_lower_case")
               .rename(columns=lambda x: re.sub(r"[^0-9a-z]+", "_", x.lower()).strip("_"))
              .pipe(log_step, "01_snake_case column names"))
```

```python
# save to interim
save_path = PROJECT_ROOT / "data" / "interim" / "pharma_sales_structure_tidiness_fixed.csv"
df.to_csv(save_path, index=False)
```
