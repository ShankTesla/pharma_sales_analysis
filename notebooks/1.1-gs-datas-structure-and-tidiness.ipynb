```python
# import libraries 

import pandas as pd
import regex
```

```python
# load data into df

df = pd.read_csv("../data/raw/salesdaily.csv", encoding="utf-8", na_values=["", "NA", "-", "?"])
```

```python
# standardise column names

df.columns = (df.columns.str.strip().str.lower()
              .str.replace(r"[^0-9a-z]+", "_", regex=True).str.strip("_"))
```

```python
# save to interim

df.to_csv('../data/interim/pharma_sales_structure_tidiness_fixed.csv', index=False)
```
