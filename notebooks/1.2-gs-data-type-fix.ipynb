```python
#import necessary libraries

import pandas as pd
import datetime
```

```python
# load data into df

df = pd.read_csv("../data/raw/salesdaily.csv", encoding="utf-8", na_values=["", "NA", "-", "?"])
```

```python
# print the first few rowws of the datum column

print(df["datum"].head())
```

```python
# check the dtype of the datum colum

print(df["datum"].dtype)
```

```python
# create a new column, date_parsed, with the parsed dates

df["datum_parsed"] = pd.to_datetime(df["datum"], format="%m/%d/%Y", errors="coerce", utc=True)
```

```python
# print the first few rows

df["datum_parsed"].head()
```

```python
# save to interim

df.to_csv('../data/interim/pharma_sales_data_type_fixed.csv', index=False)
```
